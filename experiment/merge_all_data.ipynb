```python
import pandas as pd

T6_KEPLER = pd.read_csv('kepler_Neuronet_20250805_090528.csv')
T6_KEPLER['_time'] = pd.to_datetime(T6_KEPLER['_time'])
T5_K8S = pd.read_csv('k8s_Neuronet_20250805_090507.csv')
T5_K8S['_time'] = pd.to_datetime(T5_K8S['_time'])
T4_SCA_VM = pd.read_csv('scaphandre_neuronet_20250805_090241.csv')
T4_SCA_VM['_time'] = pd.to_datetime(T4_SCA_VM['_time'])
T3_SCA_HT_3 = pd.read_csv('scaphandre_flux-node3_20250805_090248.csv')
T3_SCA_HT_4 = pd.read_csv('scaphandre_flux-node4_20250805_090248.csv')
T3_SCA_HT_5 = pd.read_csv('scaphandre_flux-node5_20250805_090249.csv')
T3_SCA_HT = pd.concat([T3_SCA_HT_3, T3_SCA_HT_4, T3_SCA_HT_5])
T3_SCA_HT['_time'] = pd.to_datetime(T3_SCA_HT['_time'])
T2_PDU_1 = pd.read_csv('pdu_flux-node1_20250805_090211.csv')
T2_PDU_3 = pd.read_csv('pdu_flux-node3_20250805_090214.csv')
T2_PDU_4 = pd.read_csv('pdu_flux-node4_20250805_090217.csv')
T2_PDU_5 = pd.read_csv('pdu_flux-node5_20250805_090219.csv')
T2_PDU = pd.concat([T2_PDU_1, T2_PDU_3, T2_PDU_4, T2_PDU_5])
T2_PDU['_time'] = pd.to_datetime(T2_PDU['_time'])
T1_PROXMOX = pd.read_csv('proxmox_neuronet_20250805_090202.csv')
T1_PROXMOX['_time'] = pd.to_datetime(T1_PROXMOX['_time'])
```

```python
# Merge T6_KEPLER with T5_K8S on time and pod_name
df_step1 = pd.merge_asof(T6_KEPLER.sort_values('_time'), T5_K8S.sort_values('_time'), on='_time', by='pod_name', direction='nearest')
print(df_step1.shape)
df_step1.head()
```

```python
# Merge df with T4_SCA_VM on time and node_name (in T4_SCA_VM node_name is named vm_name)
df_step2 = pd.merge_asof(df_step1.sort_values('_time'), T4_SCA_VM.sort_values('_time'), left_on='_time', right_on='_time', left_by='node_name', right_by='vm_name', direction='nearest')
print(df_step2.shape)
df_step2.head()
```

```python
# merge df with T3_SCA_HT on time and url on both sides
df_step3 = pd.merge_asof(df_step2.sort_values('_time'), T3_SCA_HT.sort_values('_time'), left_on='_time', right_on='_time', left_by='url_y', right_by='url', direction='nearest', suffixes=('_step2', '_sca'))
print(df_step3.shape)
df_step3.head()
```

```python
df_step3['node'] = df_step3['url'].str.extract(r'http://(flux-node\d+)')
df_step3.head()
```

```python
# merge df with T2_PDU on time and inventory server id and node in df_step3
df_step4 = pd.merge_asof(df_step3.sort_values('_time'), T2_PDU.sort_values('_time'), left_on='_time', right_on='_time', left_by='node', right_by='inventory-server-id', direction='nearest', suffixes=('_step3', '_pdu'))
print(df_step4.shape)
df_step4.head()
```
